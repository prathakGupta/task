```python
def load_boston():
  data_url = "http://lib.stat.cmu.edu/datasets/boston"
  raw_df = pd.read_csv(data_url, sep=r"\s+", skiprows=22, header=None)
  data = np.hstack([raw_df.values[::2, :], raw_df.values[1::2, :2]])
  target = raw_df.values[1::2, 2]
  return {'data': data, 'target': target}

boston = load_boston()
X = boston['data']
y = boston['target']

feature_names = [
"CRIM", "ZN", "INDUS", "CHAS", "NOX", "RM", "AGE",
"DIS", "RAD", "TAX", "PTRATIO", "B", "LSTAT"
]

w_full, residuals, rank, s = np.linalg.lstsq(X, y, rcond=None)
print("Full least-squares weight vector w (feature order):")
for name, coeff in zip(feature_names, w_full):
  print(f"{name:7s} : {coeff: .6f}")

print("\nInterpretation (based on sign):")
for name, coeff in zip(feature_names, w_full):
  direction = 'directly' if coeff>0 else ('inversely' if coeff<0 else 'no clear')
  print(f"{name:7s}: The house price is {direction} proportional to {name} (weight={coeff:.4f}).")

X_with_intercept = np.hstack([np.ones((X.shape[0],1)), X])
w_int, *_ = np.linalg.lstsq(X_with_intercept, y, rcond=None)
print('\nIntercept + coefficients (with intercept added):')
print(f"Intercept: {w_int[0]:.6f}")
for name, coeff in zip(feature_names, w_int[1:]):
  print(f"{name:7s} : {coeff: .6f}")
```

Question 7

```python
a = np.array([4.0, 3.0, 5.0, 0.0])
b = np.array([5.0, 2.0, 4.0, 1.0])
c = np.array([1.0, 0.0, 2.0, 3.0])

from math import sqrt


def euclidean(u, v):
  return np.linalg.norm(u - v)


d_ab = euclidean(a, b)
d_ac = euclidean(a, c)
d_bc = euclidean(b, c)
print("Euclidean distances:")
print(f"d(a,b) = {d_ab:.4f}")
print(f"d(a,c) = {d_ac:.4f}")
print(f"d(b,c) = {d_bc:.4f}")

pairs = {('a','b'):d_ab, ('a','c'):d_ac, ('b','c'):d_bc}
closest = min(pairs, key=pairs.get)
farthest = max(pairs, key=pairs.get)
print(f"Closest pair (Euclidean): {closest} with distance {pairs[closest]:.4f}")
print(f"Farthest pair (Euclidean): {farthest} with distance {pairs[farthest]:.4f}")

from numpy.linalg import norm

def cosine_similarity(u, v):
  if norm(u)==0 or norm(v)==0:
    return 0.0
  return np.dot(u, v) / (norm(u)*norm(v))


sim_ab = cosine_similarity(a,b)
sim_ac = cosine_similarity(a,c)
sim_bc = cosine_similarity(b,c)


cos_ab = 1 - sim_ab
cos_ac = 1 - sim_ac
cos_bc = 1 - sim_bc
print('\nCosine distances (1 - cosine_similarity):')
print(f"cos_dist(a,b) = {cos_ab:.4f}")
print(f"cos_dist(a,c) = {cos_ac:.4f}")
print(f"cos_dist(b,c) = {cos_bc:.4f}")


pairs_cos = {('a','b'):cos_ab, ('a','c'):cos_ac, ('b','c'):cos_bc}
closest_cos = min(pairs_cos, key=pairs_cos.get)
farthest_cos = max(pairs_cos, key=pairs_cos.get)
print(f"Closest pair (Cosine): {closest_cos} with distance {pairs_cos[closest_cos]:.4f}")
print(f"Farthest pair (Cosine): {farthest_cos} with distance {pairs_cos[farthest_cos]:.4f}")
```

Question 11

```python
x, k, a, b = sp.symbols('x k a b')

fA = x**3 - 1
fA_diff = sp.diff(fA, x)

fB = (-3*k * sp.log(x)) / sp.log(2)
fB_diff = sp.diff(fB, x)


fC = sp.exp(a * x**b)

fC_diff = sp.diff(fC, x)

print("Q11 A: f(x) = x^3 - 1")
print("f'(x) =", fA_diff)

print("\nQ11 B: interpreted as f(x) = log_2(x^{-3k}) = (-3k * ln x) / ln 2")
print("f(x) (symbolic) =", fB)
print("f'(x) =", sp.simplify(fB_diff))

print("\nQ11 C: f(x) = exp(a * x^b)")
print("f'(x) =", sp.simplify(fC_diff))
```
